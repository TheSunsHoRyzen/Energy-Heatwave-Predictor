```python
# Cell 1: Imports & Load Data
import pandas as pd
import matplotlib.pyplot as plt

# adjust path if needed
df_new = pd.read_csv('merged_daily_demand_weather_energy.csv')
```

```python
# Cell 2: Inspect Data
# show first few rows
df_new.head()
# show data types
df_new.dtypes
```

```python
# Cell 3: Parse Dates & Fill Missing
# ensure 'date' is datetime
df_new['date'] = pd.to_datetime(df_new['date'])

# forward-fill any gaps in Humidity (HU) and Cloud Cover (CC)
df_new[['HU', 'CC']] = df_new[['HU', 'CC']].ffill()

# optional: set date as index
# df_new.set_index('date', inplace=True)
```

```python
# Cell 4: Scatter Plots of Consumption vs. Weather Features
features = ['TX','TN','TG','RR','SS','QQ','PP','HU','CC']

for feat in features:
    plt.figure()
    plt.scatter(df_new['consumption'], df_new[feat], alpha=0.5)
    plt.title(f'Consumption vs {feat}')
    plt.xlabel('Consumption')
    plt.ylabel(feat)
    plt.show()
```
